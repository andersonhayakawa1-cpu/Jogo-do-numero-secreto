````python
import time

def resumidor_de_emails(lista_de_emails):
    lista_de_resumos = []

    for numero, email in enumerate(lista_de_emails):
        time.sleep(1) # Pausa de 1 segundo para evitar sobrecarga da API
        resposta = client.models.generate_content(
            model="gemini-3.1-flash-lite",
            contents=f"""Vou te mandar o corpo de um e-mail. Quero que você o resuma em apenas 1 linha,
                        passando o intuito daquele e-mail. Segue o e-mail: {email}"""
            )
        print(f"E-mail {numero + 1}: {resposta.text}")
        lista_de_resumos.append(f"E-mail {numero + 1}: {resposta.text}")
        print("-" * 50)

    return lista_de_resumos
````

````python
lista_de_analise_de_sentimentos = []

# Processar apenas um lote menor de resenhas para evitar o erro de quota
# Altere o número (ex: 5) para ajustar o tamanho do lote conforme necessário
for review_numero, resenha in enumerate(coluna_de_reviews[:5]):
  resposta = client.models.generate_content(
      model='gemini-3.1-flash-lite',
      contents=f"""Voce ira analisar que eu te mandarei abaixo e retorne com uma analise de sentimento.
      voce deve responder apenas com uma das seguintes palavras: Positivo, Negativo ou Neutra,
      indicando o sentimento relativo aquela resenha espeficifica.

      exemplo:
      'eu adorei esse produto' -> positivo
      'gostei, mas nao é nada especial' - Neutro
      'Odiei esse produto' - negativo

      segue a resenha a ser analisada:  {resenha}
      """
      )

  lista_de_analise_de_sentimentos.append(resposta.text)
  print(f"resenha {review_numero}: '{resenha} -> sentimento: {resposta.text}'")
  print('='*100)
````

````python
df_analised_reviews = df_reviews.head(len(lista_de_analise_de_sentimentos)).copy()
df_analised_reviews['Analises de Sentimentos'] = lista_de_analise_de_sentimentos
print(df_analised_reviews)
````

````python
import json

# Prompt ajustado para ser estrito quanto ao formato de resposta
prompt_analise = f"""
Você é um analista de dados. Vou te passar uma string contendo resenhas negativas separadas por '#####'.
Quero que você encontre as categorias de reclamações e me retorne APENAS 5 array de objetos JSON (sem saudações, sem texto conversacional e sem tags de markdown).

Cada objeto deve ter exatamente estas três chaves:
1. 'resenha_original': o texto original da resenha.
2. 'resenha_pt': a tradução exata da resenha para o português.
3. 'categoria': uma única palavra em inglês que defina o tipo de reclamação (ex: Quality, Defect, Delivery, Price, Design).

Resenhas para analisar:
{resenhas_negativas_unidas}
"""

# Executa a chamada no modelo do Gemini
resposta_gemini = client.models.generate_content(
    model='gemini-3.1-flash-lite',
    contents=prompt_analise
)

texto_resposta = resposta_gemini.text.strip()

# Limpa possíveis blocos de código markdown que a IA possa ter gerado de forma invasiva
if texto_resposta.startswith("```json"):
    texto_resposta = texto_resposta.split("```json")[1].split("```")[0].strip()
elif texto_resposta.startswith("```"):
    texto_resposta = texto_resposta.split("```")[1].split("```")[0].strip()

try:
    # Converte a string limpa em um dicionário estruturado do Python
    json_de_resenhas_classificadas = json.loads(texto_resposta)
    print("JSON carregado com sucesso!\n")

    # Transforma o JSON estruturado direto em um DataFrame para melhor visualização
    df_resultado_final = pd.DataFrame(json_de_resenhas_classificadas)
    display(df_resultado_final)

except json.JSONDecodeError as e:
    print("Erro ao decodificar. O modelo não obedeceu o formato JSON e retornou texto puro:")
    print(texto_resposta)
````
